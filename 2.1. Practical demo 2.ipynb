#### A. Exploring UniProt

1. Go to [UniProt](https://www.uniprot.org) and look for a **reviewed entry** for the human glutathione S-transferase protein (A3 isoform).
    - Use this query: **"glutathione s-transferase A3"**, and click on both the **"reviewed"** widget (golden logo) on the sidebar on the left to get the curated record for this protein
    - Click on **"human"** to narrow down your search to humans
    - Questions:
      - What is the sequence length of GSTA3?
      - From the humans, how many GSTA3 entries were not reviewed (i.e. automatically annotated)?

2. Click on the accession **"Q16772"** from the table of results. This will show you valuable information as annotations to the human GSTA3. These are gathered from various cross-linked databases 
    - E.g. Gene name, function, sub-cellular localisation, variations, etc
    - Questions
        - How many natural variations are reported for human GST3A?
        - From the variant viewer, where is the first "pathogenic mutation" located?
            - What kind of mutation is it?

#### B. _Ab initio_ 3D structure modelling using AlphaFold

1. Go to the sequence section, and copy the sequence (click "Copy sequence") and paste it in a text editor (e.g. Notepad). 
   - Mutate residue 10 from phenylalanine to proline using your text editor
   - Go to the **[alphafold server](https://alphafoldserver.com)** to construct an **_ab initio_** 3D model of your mutant's 3D structure  
      - Clear the page if there is any unrelated material cluttering the page
      - Leave "Entity type" to "Protein"
      - Change "Copies" to 2, as we're dealing with a homodimer
      - Paste your mutant sequence in the textbox, and press "Continue and preview job", and confirm 
      - Click on your job once it's done and download the results
         - Note the confidence levels for the different segments of your _ab initio_ model

#### C. Using the ZINC database for small molecule (ligand) retrieval

1. Go to the ZINC database and search for the compound "glutathione". 
    - Download it to your computer in SDF format (Note its name and where you saved it).
    - Note the absence of all hydrogen atoms from the molecule
        - In a real setting, some of these should be present, but we'll not do so in this demonstration

#### D. Dock the glutathione to the mutant GST3A

1. Go to the [CB-Dock 2 server](https://cadd.labshare.cn/cb-dock2/), and click on "Get Started >>"
    - Select a server that says "Available" if the notification window appears
    - Upload your mutant GST3A (which will be your receptor)
        - Open one of your models (in CIF format) using PyMOL, then in its prompt, write "save receptor.pdb", making sure you know where it's saved
        - If PyMOL is not installed, [Click here](https://pymol.org/#download) to download the "Windows EXE Installer". Follow the instructions for installing for the user (you)
    - Upload your ligand sdf file (glutathione)
    - Wait for the files to be uploaded, and click "Auto Blind Docking", then click on "view results"
    - Click on "Show" in the "Structure-based Blind Docking" section of the results page
    - Questions:
        - What is the highest affinity bound pose for glutathione, and what is it's Vina binding energy score (kcal/mol)?
        - Is it binding the protein at the same region as the WT (unmutated) protein?
        - What are the residues contacting glutathione in the mutant?
